```python
import pandas as pd
```

```python
df= pd.read_csv("/kaggle/input/datasets/rohankashyap5526/quots-data/qoute_dataset.csv")
```

```python
q1=df['quote']
```

```python
q1[0]
```

```python
q1[0][0]
```

```python
q2=q1.apply(lambda x: x.replace(q1[0][0], "").replace(q1[0][-1], ""))
```

```python
q2[0]
```

```python
q3= q2.apply(lambda x: x.split())
```

```python
q3
```

```python
input=[]
for i in q3:
    # print(i)
    for j in range(1, len(i)):
        input.append(i[:j])
        
    # break
```

```python
# input
```

```python
input_string=[]
for seq in input:
    word= " ".join(seq)
    input_string.append(word)
```

```python
len(input_string)
```

```python
output= []
for i in range(len(input_string)):
    word=  input_string[i].split()
    # print(word[-1])
    output.append(word[-1])
    # break
    
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts(input_string)
```

```python
vocab_size=len(tokenizer.word_index)
```

```python
# tokenizer.word_index
```

```python
X= input_string
y= output
```

```python
X_seq= tokenizer.texts_to_sequences(X)
y_seq= tokenizer.texts_to_sequences(y)
```

```python
X_seq[:10]
```

```python
y_seq[:10]
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
max_len=0
for i in X_seq:
    if len(i)>max_len:
        max_len=len(i)
```

```python
max_len
```

```python
X_pad= pad_sequences(X_seq, maxlen= max_len, padding='pre')
```

```python
X_pad[:10]
```

```python
y_seq[:10]
```

```python
import numpy as np
```

```python
# np.array(y_seq)
```

```python
out=[]
for i in y_seq:
    # print(i)
    if len(i)>0:
        out.append(i[0])
    else:
        out.append(0)
    
```

```python
out[:10]
```

```python
len(out)
```

```python
len(X_pad)
```

```python
y_out= np.array(out)
```

```python
y_out[:10]
```

```python
from tensorflow.keras.utils import to_categorical
```

```python

y_cat= to_categorical(y_out, num_classes=vocab_size+1 )
```

```python
y_cat[:10]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, LSTM, GRU

```

```python
emd_dim= 50
rnn_units= 64
```

```python
max_len
```

```python
model= Sequential(
    [
        Embedding(input_dim= vocab_size+1, output_dim= emd_dim, input_length= max_len),
        LSTM(rnn_units),
        Dense(vocab_size+1, activation='softmax')
    ]
)
```

```python
model.compile(loss='categorical_crossentropy', optimizer="adam", metrics=['accuracy'])
```

```python
model.fit(X_pad, y_cat, epochs=5, verbose=1)
```
